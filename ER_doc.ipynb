```python
import ER_doc
import matplotlib.pyplot as plt
import numpy
import scipy.stats as st
import random
import seaborn as sns
```

```python
def run_sim(G, max_iteration):
    num_iteration = 0
    opinion_history = []
    while num_iteration < max_iteration and not G.consensus_reached(0.001) and not G.opinions_stabilized(0.0000001):
    # while num_iteration < max_iteration and not G.consensus_reached(0.001):
        for node in G.graph.nodes():
            neighbors = list(G.graph.neighbors(node))
            weighted_sum = sum(G.weight_matrix[node][n] * G.agents[n].opinion for n in neighbors)
            new_opinion = weighted_sum / sum(G.weight_matrix[node][n] for n in neighbors)
            G.agents[node].opinion = new_opinion
        num_iteration += 1
        opinions_snapshot = [G.agents[node].opinion for node in G.graph.nodes()]
        opinion_history.append(opinions_snapshot)
    num_nodes = G.num_of_nodes
    time_steps = range(num_iteration)
    return opinion_history, num_nodes, num_iteration
```

```python
def count_extreme_opinions(opinions):
    above_0_8 = sum(1 for opinion in opinions if opinion > 0.8)
    below_0_2 = sum(1 for opinion in opinions if opinion < 0.2)
    return above_0_8, below_0_2
```

```python
def heat_map_plot(heatmap_data, n_bot, n_titled_bots_per):
    plt.figure(figsize=(8, 6))
    sns.heatmap(heatmap_data, xticklabels=n_titled_bots_per, yticklabels=n_bot, 
                cmap='viridis', annot=True, fmt=".3f", cbar_kws={'label': '% Extreme Opinions'})

    plt.xlabel('Proportion of Bots with Titles')
    plt.ylabel('Number of Bots')
    plt.title('Heatmap of Extreme Opinions')
    plt.tight_layout()
    plt.show()
```

```python
# Pad each list to the same length (e.g., max time step)
def pad_sequences(seq_list, pad_value=numpy.nan):
    max_len = max(len(seq) for seq in seq_list)
    padded = numpy.full((len(seq_list), max_len), pad_value)
    for i, seq in enumerate(seq_list):
        padded[i, :len(seq)] = seq
    return padded

# Process each percentage group
data_0 = pad_sequences(E1_aveopinion_0)
mean_0 = numpy.nanmean(data_0, axis=0)
std_0 = numpy.nanstd(data_0, axis=0)

data_20 = pad_sequences(E1_aveopinion_20)
mean_20 = numpy.nanmean(data_20, axis=0)
std_20 = numpy.nanstd(data_20, axis=0)

data_40 = pad_sequences(E1_aveopinion_40)
mean_40 = numpy.nanmean(data_40, axis=0)
std_40 = numpy.nanstd(data_40, axis=0)

data_60 = pad_sequences(E1_aveopinion_60)
mean_60 = numpy.nanmean(data_60, axis=0)
std_60 = numpy.nanstd(data_60, axis=0)

data_80 = pad_sequences(E1_aveopinion_80)
mean_80 = numpy.nanmean(data_80, axis=0)
std_80 = numpy.nanstd(data_80, axis=0)

data_100 = pad_sequences(E1_aveopinion_100)
mean_100 = numpy.nanmean(data_100, axis=0)
std_100 = numpy.nanstd(data_100, axis=0)

# Plot
plt.figure(figsize=(12, 8))

time_0 = numpy.arange(len(mean_0))
plt.plot(time_0, mean_0, color='orange', label="0 doctors with 'Dr.'")
plt.fill_between(time_0, mean_0 - std_0, mean_0 + std_0, color='orange', alpha=0.2)

time_20 = numpy.arange(len(mean_20))
plt.plot(time_20, mean_20, color='pink', label="20 doctors with 'Dr.'")
plt.fill_between(time_20, mean_20 - std_20, mean_20 + std_20, color='pink', alpha=0.2)

time_40 = numpy.arange(len(mean_40))
plt.plot(time_40, mean_40, color='yellow', label="40 doctors with 'Dr.'")
plt.fill_between(time_40, mean_40 - std_40, mean_40 + std_40, color='yellow', alpha=0.2)

time_60 = numpy.arange(len(mean_60))
plt.plot(time_60, mean_60, color='lightblue', label="60 doctors with 'Dr.'")
plt.fill_between(time_60, mean_60 - std_60, mean_60 + std_60, color='lightblue', alpha=0.2)

time_80 = numpy.arange(len(mean_80))
plt.plot(time_80, mean_80, color='lightgreen', label="80 doctors with 'Dr.'")
plt.fill_between(time_80, mean_80 - std_80, mean_80 + std_80, color='lightgreen', alpha=0.2)

time_100 = numpy.arange(len(mean_100))
plt.plot(time_100, mean_100, color='purple', label="100 doctors with 'Dr.'")
plt.fill_between(time_100, mean_100 - std_100, mean_100 + std_100, color='purple', alpha=0.2)

plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_0]), color='grey', linestyle='dashed')
plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_20]), color='grey', linestyle='dashed')
plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_40]), color='grey', linestyle='dashed')
plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_60]), color='grey', linestyle='dashed')
plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_80]), color='grey', linestyle='dashed')
plt.axvline(x=numpy.nanmean([len(x) for x in E1_aveopinion_100]), color='grey', linestyle='dashed')

plt.axhline(y=1, color='grey', linestyle='dashed')
plt.xlabel('Time Steps', fontsize=25)
plt.ylabel('Opinion', fontsize=25)
# plt.title('E1 - Average Opinion Over Time on the ER Network', fontsize=25)
plt.xticks(fontsize=20)
plt.yticks(fontsize=25)
plt.legend(fontsize=25)
plt.show()
```

```python
import numpy as numpy
import matplotlib.pyplot as plt
import pandas as pd

def analyze_distribution(numbers, num_bins=10):
    """
    Analyze and visualize the distribution of numbers between 0 and 1
    
    Parameters:
    numbers: list or array of numbers between 0 and 1
    num_bins: number of intervals to divide the range into
    """
    
    # Convert to numpy array for easier handling
    data = numpy.array(numbers)
    
    # Basic statistics
    print("=== DISTRIBUTION ANALYSIS ===")
    print(f"Total numbers: {len(data)}")
    print(f"Min value: {data.min():.4f}")
    print(f"Max value: {data.max():.4f}")
    print(f"Mean: {data.mean():.4f}")
    print(f"Standard deviation: {data.std():.4f}")
    print(f"Median: {numpy.median(data):.4f}")
    print()
    
    # Create bins and count frequencies
    bin_edges = numpy.linspace(0, 1, num_bins + 1)
    counts, _ = numpy.histogram(data, bins=bin_edges)
    
    # Create interval labels
    intervals = []
    for i in range(len(bin_edges) - 1):
        intervals.append(f"[{bin_edges[i]:.2f}, {bin_edges[i+1]:.2f})")
    
    # Print interval counts
    print("=== INTERVAL COUNTS ===")
    for interval, count in zip(intervals, counts):
        percentage = (count / len(data)) * 100
        print(f"{interval}: {count} points ({percentage:.1f}%)")
    print()
    
    # Create visualizations
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 10))
    
    # 1. Histogram
    ax1.hist(data, bins=bin_edges, alpha=0.7, color='skyblue', edgecolor='black')
    ax1.set_title('Final Opinion Distribution')
    ax1.set_xlabel('Value')
    ax1.set_ylabel('Frequency')
    ax1.grid(True, alpha=0.3)
    
    # 2. Bar chart of intervals
    ax2.bar(range(len(intervals)), counts, color='lightcoral', alpha=0.7)
    ax2.set_title('Count by Interval')
    ax2.set_xlabel('Interval')
    ax2.set_ylabel('Count')
    ax2.set_xticks(range(len(intervals)))
    ax2.set_xticklabels([f"[{bin_edges[i]:.1f},{bin_edges[i+1]:.1f})" 
                        for i in range(len(bin_edges)-1)], rotation=45)
    ax2.grid(True, alpha=0.3)
    
    # 3. Cumulative distribution
    sorted_data = numpy.sort(data)
    y_values = numpy.arange(1, len(sorted_data) + 1) / len(sorted_data)
    ax3.plot(sorted_data, y_values, color='green', linewidth=2)
    ax3.set_title('Cumulative Distribution Function')
    ax3.set_xlabel('Value')
    ax3.set_ylabel('Cumulative Probability')
    ax3.grid(True, alpha=0.3)
    
    # 4. Box plot
    ax4.boxplot(data, vert=True)
    ax4.set_title('Box Plot')
    ax4.set_ylabel('Value')
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Return summary dataframe
    summary_df = pd.DataFrame({
        'Interval': intervals,
        'Count': counts,
        'Percentage': (counts / len(data)) * 100
    })
    
    return summary_df

# Example usage:
if __name__ == "__main__":
    
    # Analyze the distribution
    print("Analyzing example data...")
    summary = analyze_distribution(opinion_history[499], num_bins=10)
    
    print("\n=== SUMMARY TABLE ===")
    print(summary.to_string(index=False))
    print("\n" + "="*50)
    print("Analysis with 5 bins:")
    summary_5bins = analyze_distribution(opinion_history[499], num_bins=5)
```
